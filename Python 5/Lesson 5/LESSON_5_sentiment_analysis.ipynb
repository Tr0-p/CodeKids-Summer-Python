````python
# Import libraries
import requests
import pandas as pd
from sklearn.metrics import accuracy_score
````

````python
# https://machinelearningforkids.co.uk/
# This function will pass your text to the machine learning model
# and return the top result with the highest confidence
def classify(text):
    key = "8176a890-2757-11ef-832e-2136c40f26ea97269a44-4220-47e8-bffd-20181e5834cf"
    url = "https://machinelearningforkids.co.uk/api/scratch/"+ key + "/classify"

    response = requests.get(url, params={ "data" : text })

    if response.ok:
        responseData = response.json()
        topMatch = responseData[0]
        return topMatch
    else:
        response.raise_for_status()

# This function will predict the sentiment label of many examples
# Examples are provided in a dataframe (data)
# The function returns a list of predicted labels
def predict(data):
  pred_labels = []
  for text in data['example']:
    result = classify(text)
    label = result['class_name']
    pred_labels.append(label)

  return pred_labels
````

### Test model on 1 Example

````python
# CHANGE THIS to something you want your machine learning model to classify
text = "I'm so sad."
demo = classify(text)

label = demo["class_name"]
confidence = demo["confidence"]

# CHANGE THIS to do something different with the result
print("result: '%s' with %d%% confidence" % (label, confidence))
````

### Test model on many examples (test set)

````python
# Path to your CSV file
file_path = "/content/sentiment_test_examples.csv"
data = pd.read_csv(file_path)

data.head()
````

````python
# There are 6 sentiments in our test set
print(data['class_name'].value_counts())
````

### Evaluate model performance using the 'accuracy' metric

````python
# Model 1 (5 examples)

# predicted labels
pred_labels = predict(data)

# true labels
true_labels = data['class_name'].tolist()

# calculate accuracy
accuracy = accuracy_score(true_labels, pred_labels)
print(accuracy)
````

* An accuracy of 0 means NONE of the sentiments were predicted correctly
* An accuracy of 1 means ALL of the sentiments were predicted correctly
