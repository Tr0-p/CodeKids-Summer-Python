### Making a Calculator


#### 1. Create the functions for the calculator operations
FUNCTIONS: Are like packaged pieces of instructions that we can
call over and over again. For example, print() and input() are functions

Let's make our own functions!

```python
# Making a new functions we use a word called def
# Create the functions for a simple calculator

# Create a function for adding: +
def add(num1, num2):
    total = num1 + num2
    print(f"Adding {num1} and {num2}. Answer: {total} ")

# Create a function for subtracting: -
def subtract(num1, num2):
    total = num1 - num2
    print(f"Subtracting {num1} and {num2}. Answer: {total} ")
```

```python
# Let's call the function - this is the ADDITION function
add(35, 139)
```

```python
# Calling the subtract function and giving it 2 numbers
subtract(7, 1)
```

```python
## CREATE THE THE REST OF YOUR FUNCTIONS
# multiply: *
# divide: /

def multiply(num1, num2):
    total = num1 * num2
    print(f"Multiplying {num1} and {num2}. Answer: {total} ")

def divide(num1, num2):
    total = num1 / num2
    print(f"Dividing {num1} and {num2}. Answer: {total} ")
```

```python
multiply(7, 5)
```

```python
divide(100, 10)
```

#### 2. Create a program that will ask the user for 2 numbers and an operator
The program will perform the operation and print the result

```python
# Create a forever loop
while True:

    # Ask user for the input operator
    operator = input("Do you want to +, -, * or /: ")

    # Ask user for the input numbers
    # We have to convert input() to integers
    number1 = int(input("Enter the first number: "))
    number2 = int(input("Enter the second number: "))

    # CREATE IF/ELSE statements based on operator condition
    if operator == "+":
      add(number1, number2)
      break

    # CREATE YOUR ELIFs for subtract, multiply, and divide

    elif operator == "-":
      subtract(number1, number2)
      break

    elif operator == "*":
      multiply(number1, number2)
      break

    elif operator == "/":
      divide(number1, number2)
      break

    else:
      print("Sorry, I don't know this operator. ")
      print("Please enter: +, -, * or / ")
```
