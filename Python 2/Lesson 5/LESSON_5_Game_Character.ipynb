```python
# Let's gather some user information
def inputs():
   name = input("What is your character's name?\n")
   age = int(input("How old is your character?\n"))
   weapon = input("What weapon do they use?\n")
   pet = input("What pet do they have?\n")

   strength = int(input("How strong are they out of 100?\n"))
   speed = int(input("How fast are they out of 100?\n"))
   intelligence = int(input("How smart are they out of 100?\n"))

# Let's print the information we gathered
def outputs():
   print(f"This is {name}. They are {age} years old.")
   print(f"When they fight they use the {weapon}.")
   print(f"When they aren't fighting they are looking after their {pet}")

   print(f"Strength: {strength}")
   print(f"Speed: {speed}")
   print(f"Intelligence: {intelligence}")

# Let's write a main function to run the program
# First we will run the inputs function, then the outputs function
def main():
  inputs()
  outputs()

# Let's run the main function
main()
```

How can we store the information from one function so that we can use it in another function? How can we transport variables between functions? --> We use RETURN


```python
# Let's just run the inputs function again and return the 'name' variable
def inputs():
   name = input("What is your character's name?\n")

   return name

name = inputs()
print(name)
```

How can we transport more than just one variable? --> in a DICTIONARY

```python
# For examples this is a dictionary that stores name and age of our character
my_dict = {'name': 'Steve', 'age': 2}
my_dict['name']
```

Let's use this knowledge to fix our character builder code!

```python
# Let's gather some user information
def inputs():
  dictionary = {}  # create an empty dictionary
  dictionary['name'] = input("What is your character's name?\n")
  dictionary['age'] = int(input("How old is your character?\n"))
  dictionary['weapon'] = input("What weapon do they use?\n")
  dictionary['pet'] = input("What pet do they have?\n")

  dictionary['strength'] = int(input("How strong are they out of 100?\n"))
  dictionary['speed'] = int(input("How fast are they out of 100?\n"))
  dictionary['intelligence'] = int(input("How smart are they out of 100?\n"))

  # How can we return all variables?
  return dictionary

# Let's transfer the information from the inputs function to the outputs function
def outputs(dictionary):
  print(f"This is {dictionary['name']}. They are {dictionary['age']} years old.")
  print(f"When they fight they use the {dictionary['weapon']}.")
  print(f"When they aren't fighting they are looking after their {dictionary['pet']}")

  print(f"Strength: {dictionary['strength']}")
  print(f"Speed: {dictionary['speed']}")
  print(f"Intelligence: {dictionary['intelligence']}")

# Let's write a main function to run the program
# First we will run the inputs function, then the outputs function
def main():
  my_dict = inputs()
  outputs(my_dict)

# Let's run the main function
main()
```
