This project connects to a service which returns the population (or other facts) of countries of the world. We will learn about APIs and use the REQUESTS module to connect to URLs.

* **API:** It's like a magic box on the internet that gives us information when we ask nicely. Here, we're asking for the weather.
* **Function:** It's like a little helper that knows how to do a specific task. We have a function to get weather information from the internet, and one frunction to format the information.

```python
# We start by importing the requests library.
# This library helps us send requests to websites and get information back.
import requests

# This function extracts information (temp, conditions) from a json object
def format_response(weather, units):

    city = weather['name']
    desc = weather['weather'][0]['description']
    temp = weather['main']['temp']

    final_text = f'City: {city} \nConditions: {desc} \nTemperature: {temp}'

    return final_text


# This function takes a city name as input and returns the weather there.
def get_weather(city, units):
    weather_key = 'a4aa5e3d83ffefaba8c00284de6ef7c3'  # API Key: Our unique key to allow us to request stuff.
    url = 'https://api.openweathermap.org/data/2.5/weather'  # URL: Where we are requesting from.

    params = {'APPID': weather_key, 'q': city, 'units': units}  # What we are sending off to the server.
    response = requests.get(url, params=params)  # Ask the server above for the information we need.
    weather = response.json()  # Get the response in a readable format.
    formatted_weather = format_response(weather, units)  # Store the formatted weather results.

    return formatted_weather


def main():
    while True:
        city = input("City: ")
        units = input("Metric or Imperial? ").lower()

        print(get_weather(city, units))


main()
```
