```python
def choose_door(floor_description):
    return input(f"You are on the {floor_description}. Choose a door: ").strip()

### FLOOR 1 ###
def floor_one(inventory, hearts):
    print("\nFloor 1: You see three doors labeled A, B, C.")
    choice = choose_door("first floor")
    status = None

    if choice.upper() == 'A':
        print("You found a rusty key!")
        inventory.append('rusty key')
    elif choice.upper() == 'B':
        if 'rusty key' in inventory:
            print("You opened the door with the rusty key!")
            print("You found a staircase!")
            status = 'next_floor'
        else:
            print("The door is locked. You need a rusty key.")
    elif choice.upper() == 'C':
        print("A trapdoor opens beneath you and you fall into a pit! Start over!")
        hearts -= 1
        status = 'start_over'
    else:
        print(f"Door {choice} doesn't exist!")

    return inventory, hearts, status

### FLOOR 2 ###
def floor_two(inventory, hearts):
    print("\nFloor 2: You see three doors colored blue, red, green.")
    choice = choose_door("second floor")
    status = None

    if choice.lower() == 'blue':
        if 'silver key' in inventory:
            print("You opened the door with the silver key!")
            print("You found a staircase!")
            print("You found a clue! It says, 'The golden key is under door 1.'")
            status = 'next_floor'
        else:
            print("The door is locked. You need a silver key.")
    elif choice.lower() == 'red':
        print("As you grab the door handle, it shocks you with a surge of electricity! Start over!")
        hearts -= 1
        status = 'start_over'
    elif choice.lower() == 'green':
        print("You found a silver key!")
        inventory.append('silver key')
    else:
        print(f"Door {choice} doesn't exist!")

    return inventory, hearts, status

### FLOOR 3 ###
def floor_three(inventory, hearts):
    print("\nFloor 3: You see doors numbered 1, 2, 3.")
    choice = choose_door("third floor")
    status = None

    if choice == '1':
        print("You open the door only to disturb a nest of venomous snakes! Start over!")
        hearts -= 1
        status = 'start_over'
    elif choice == '2':
        print("You found a golden key!")
        inventory.append('golden key')
    elif choice == '3':
        if 'golden key' in inventory:
            print("You opened the door with the golden key. Congratulations, you escaped the building!")
            status = 'escaped'
        else:
            print("A small room with a window showing the path to freedom. You need a golden key!")
    else:
        print(f"Door {choice} doesn't exist!")

    return inventory, hearts, status

### MOVE FLOORS ###
def move_player(current_floor, status):
    if status == 'next_floor':
        new_floor = current_floor + 1
    elif status == 'start_over':
        new_floor = 1
    else:
        new_floor = current_floor
    return new_floor


### GAME START ###

def play_game():
    print("Welcome to the Multi-Floor Maze Game!")
    print("You are standing at the entrance of a mysterious building with many floors. Your goal is to escape.")
    name = input("Enter your name: ")
    print(f"Good luck, {name}!")

    inventory = []  # for collecting keys or other items
    current_floor = 1  # to keep track of which floor we are on
    hearts = 3

    while hearts > 0:
        print(f"\n{name}, you have {hearts} hearts remaining.")
        if current_floor == 1:
            inventory, hearts, status = floor_one(inventory, hearts)
        elif current_floor == 2:
            inventory, hearts, status = floor_two(inventory, hearts)
        elif current_floor == 3:
            inventory, hearts, status = floor_three(inventory, hearts)

        if status == 'escaped':
            print(f"\nCongratulations, {name}, you found the exit and won the game!")
            break

        current_floor = move_player(current_floor, status)

    if hearts <= 0:
        print(f"\nSorry, {name}, you have run out of hearts. Game over!")

play_game()
```
