### Creating a Chatbot

```python
# Function to start the conversation and gather basic information about the user
def startup():
   # Asking the user's name
   name = input("What's your name?\n")

   # Asking the user's age and converting it to a number
   age = int(input(f"Hello {name}, how old are you?\n"))

   # Returning the user's name and age
   return name, age


# Function to ask questions and interact with the user based on their responses
def questions(name):
  # Asking the user if they want a drink
  wants_drink = input(f"Can I get you a drink, {name}?\n")
  # Checking if the user wants a drink
  if wants_drink.lower() == "yes":
    print("Okay, let me get you one!")

    # If yes, calling the function to provide a drink
    drink()


# Function to provide a drink to the user
def drink():
  # Asking the user if they want coffee
  coffee = input("Do you want coffee?\n")

  # Checking if the user wants coffee
  if coffee.lower() == "yes":
    print("Coffee is on its way")

  # If the user doesn't want coffee asking the user if they want juice or tea instead
  elif coffee.lower() == "no":
    choice = input("Would you like a juice or a tea instead?")
    # Checking the user's choice

    if choice.lower() == "juice":
      # If the user wants juice, providing juice
      print("Juice is on its way")

    elif choice.lower() == "tea":
      # If the user wants tea, providing tea
      print("I will get you a tea. Just give me one sec!")

    else:
      # If the user's response is not understood, apologizing
      print("Sorry, I didn't get that.")


# Main function to start the program
def main():
   # Calling the startup function to gather user's name and age
   name, age = startup()
   # Calling the questions function to continue the conversation based on user's responses
   questions(name)

# Starting the program execution with main()
main()
```
